```python
import matplotlib.pyplot as plt
import numpy as np

losses = [
    1.66,1.71,1.70,1.67,1.62,1.63,1.64,1.56,1.67,
    1.63,1.59,1.63,1.63,1.63,1.56,1.60,1.61,1.54,1.65,
    1.60,1.58,1.62,1.62,1.62,1.54,1.60,1.60,1.54,1.64,
    1.60,1.57,1.61,1.61,1.61,1.54,1.59,1.60,1.53,1.63,
    1.59,1.56,1.60,1.59,1.60,1.52,1.58,1.58,1.51,1.62,
    1.58,1.55,1.58,1.58,1.59,1.50,1.57,1.57,1.50,1.61,
    1.57,1.54,1.57,1.56,1.58,1.49, 1.56, 1.56, 1.50, 1.60,
    1.56, 1.53, 1.56, 1.55, 1.57
]

epochs = np.arange(1, len(losses) + 1)

# Moving average
window = 5
moving_avg = np.convolve(losses, np.ones(window)/window, mode="valid")

plt.figure(figsize=(12, 6))

# Raw loss
plt.plot(
    epochs,
    losses,
    marker="o",
    linewidth=1,
    alpha=0.6,
    label="Mean Loss"
)

# Smoothed trend
plt.plot(
    epochs[window-1:],
    moving_avg,
    linewidth=3,
    label=f"{window}-Epoch Moving Average"
)

plt.title("Training Loss vs Epoch")
plt.xlabel("Epoch")
plt.ylabel("Mean Loss")
plt.grid(True, alpha=0.3)
plt.legend()

plt.tight_layout()
plt.savefig("loss_curve.png", dpi=300)
plt.show()
```
